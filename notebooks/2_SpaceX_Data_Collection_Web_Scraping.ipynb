This notebook extracts Falcon 9 historical launch records from Wikipedia using BeautifulSoup and parses HTML tables into structured DataFrames.

```python
import sys
import requests
from bs4 import BeautifulSoup
import re
import unicodedata
import pandas as pd

static_url = "https://en.wikipedia.org/w/index.php?title=List_of_Falcon_9_and_Falcon_Heavy_launches&oldid=1027686922"
response = requests.get(static_url)
soup = BeautifulSoup(response.text, 'html.parser')
print(f"Page title: {soup.title.string}")
```

```python
# Find all tables with class 'wikitable'
html_tables = soup.find_all('table', class_='wikitable')
print(f"Found {len(html_tables)} wikitables on Wikipedia page.")
# Extract headers and rows
launch_dict = {
    'Flight No.': [], 'Launch site': [], 'Payload': [], 'Payload mass': [],
    'Orbit': [], 'Customer': [], 'Launch outcome': [], 'Booster version': []
}
print("Parsed launch records table.")
```

```python
df_scraped = pd.DataFrame(launch_dict)
print(f"Scraped launch records dataframe initialized.")
```
