This notebook marks launch sites on an interactive Folium map, builds marker clusters for landing success/failure, and calculates proximity distances to coastlines, railways, highways, and cities.

```python
import pandas as pd
import numpy as np

spacex_df = pd.read_csv('../data/spacex_launch_geo.csv')
print("Loaded geospatial dataset:")
spacex_df[['Launch Site', 'Lat', 'Long', 'class']].head()
```

```python
# Haversine distance calculation formula
def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0 # Radius of earth in km
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = np.sin(dlat / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2)**2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))
    return R * c

# Distance from CCAFS SLC-40 to nearest coastline
dist_coast = calculate_distance(28.563197, -80.576820, 28.56367, -80.56763)
print(f"Distance to Coastline: {dist_coast:.2f} km")
```
